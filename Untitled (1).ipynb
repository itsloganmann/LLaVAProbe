```python
# ===============================
#   PROCESS VQA DATASET
# ===============================
import os
import json

annotations_path = "data/raw/v2_mscoco_train2014_annotations.json"
questions_path = "data/raw/v2_OpenEnded_mscoco_train2014_questions.json"
processed_path = "data/processed/filtered_vqa_with_links.json"

os.makedirs("data/processed", exist_ok=True)

CATEGORY_MAPPING = {
    "what is this": "object identification",
    "what color": "color recognition",
    "how many": "counting",
    "what kind": "classification",
    "where": "location",
    "who": "person identification",
    "when": "time-related",
    "why": "reasoning",
    "which": "comparison",
    "other": "other",
}

def categorize_question(question_text, answer):
    for keyword, category in CATEGORY_MAPPING.items():
        if keyword in question_text.lower():
            return category
    return "yes/no" if answer.lower() in ["yes", "no"] else "other"

# ---- LOAD DATA ----
print("📂 Loading annotations and questions...")
with open(annotations_path, "r") as f:
    annotations = json.load(f)

with open(questions_path, "r") as f:
    questions = json.load(f)

print(f"✅ Loaded {len(annotations['annotations'])} annotations and {len(questions['questions'])} questions")

# Map question_id → question text
question_text_map = {q["question_id"]: q["question"] for q in questions["questions"]}

# ---- CREATE FILTERED ENTRIES ----
filtered_data = []
for ann in annotations["annotations"][:5000]:  # first 5K for fast testing
    qid = ann["question_id"]
    img_id = str(ann["image_id"]).zfill(12)

    q_text = question_text_map.get(qid, "Unknown question")
    img_url = f"http://images.cocodataset.org/train2014/COCO_train2014_{img_id}.jpg"

    entry = {
        "image_id": img_id,
        "image_url": img_url,
        "question_id": qid,
        "question_text": q_text,
        "category": categorize_question(q_text, ann["multiple_choice_answer"]),
        "answer": ann["multiple_choice_answer"],
    }
    filtered_data.append(entry)

# ---- SAVE PROCESSED DATA ----
with open(processed_path, "w") as f:
    json.dump(filtered_data, f, indent=2)

print(f"\n✅ Processed {len(filtered_data)} entries and saved to {processed_path}")

```

```python
# ===============================
#   TEST MODEL ON ONE EXAMPLE (FAST)
# ===============================
from PIL import Image
import requests
import torch
import re

# --- Load processed VQA data ---
with open("data/processed/filtered_vqa_with_links.json", "r") as f:
    vqa_data = json.load(f)

example = vqa_data[0]
image_url = example["image_url"]
gt_answer = example["answer"]
question = example["question_text"]

print("🖼️ Image URL:", image_url)
print("❓ Question:", question)
print("✅ Ground Truth Answer:", gt_answer)

# --- Load image ---
image = Image.open(requests.get(image_url, stream=True).raw).convert("RGB")
image = image.resize((224, 224))  # smaller = faster

# --- Smart cleaning function ---
def clean_predicted_answer(pred, reference_answer=None):
    pred = pred.strip()
    # Try to find reference answer in output
    if reference_answer:
        match = re.search(r'\b' + re.escape(reference_answer) + r'\b', pred, re.IGNORECASE)
        if match:
            return match.group(0).lower()
    # Otherwise, take first word after last '?' or last line
    if '?' in pred:
        pred = pred.split('?')[-1].strip()
    if '\n' in pred:
        pred = pred.split('\n')[-1].strip()
    pred = pred.split()[0] if pred else ''
    pred = re.sub(r'[^\w]', '', pred)
    return pred.lower()

# --- Prepare input ---
prompt = f"<image>\n{question}"
inputs = processor(text=prompt, images=[image], return_tensors="pt").to("cuda")

# --- Generate answer ---
with torch.no_grad():
    output_ids = model.generate(**inputs, max_new_tokens=20)

raw_answer = processor.decode(output_ids[0], skip_special_tokens=True)
cleaned_answer = clean_predicted_answer(raw_answer, reference_answer=gt_answer)
correct = int(cleaned_answer == gt_answer.lower())

print("\n🧠 Raw Model Prediction:", raw_answer)
print("✂️ Cleaned Prediction (one word):", cleaned_answer)
print("✅ Correct?", bool(correct))
```

```python
# ===============================================
# CELL 7 — EXTRACT TOKEN CONFIDENCE FEATURES
# ===============================================
import torch, numpy as np, re
from tqdm import tqdm
from PIL import Image
import requests, json, os

NUM_SAMPLES = 1500  # More samples for better training
IMAGE_SIZE = (224, 224)

def clean_answer(pred, ref):
    """Clean predicted answer to match reference format"""
    pred = pred.strip().lower()
    ref = ref.strip().lower()
    
    # If ref is in prediction, return it
    if ref in pred:
        return ref
    
    # Otherwise extract last word
    pred = re.sub(r"[^a-z]", " ", pred).strip().split()
    return pred[-1] if pred else ""

print("📂 Loading VQA data...")
with open("data/processed/filtered_vqa_with_links.json", "r") as f:
    vqa_data = json.load(f)

features = []
skipped = 0

print(f"🔄 Processing {NUM_SAMPLES} samples...")
for i, entry in enumerate(tqdm(vqa_data[:NUM_SAMPLES], desc="Extracting features")):
    try:
        # Load image
        image = Image.open(requests.get(entry["image_url"], stream=True).raw).convert("RGB")
        image = image.resize(IMAGE_SIZE)
        
        # Prepare prompt
        prompt = f"<image>\n{entry['question_text']}"
        inputs = processor(text=prompt, images=[image], return_tensors="pt").to(device)

        with torch.no_grad():
            # Generate with token scores - THIS IS THE KEY!
            gen_output = model.generate(
                **inputs,
                max_new_tokens=20,
                return_dict_in_generate=True,
                output_scores=True,  # Get logits for each token
                do_sample=False  # Greedy decoding for consistency
            )
        
        # Decode prediction
        pred_text = processor.decode(gen_output.sequences[0], skip_special_tokens=True)
        cleaned = clean_answer(pred_text, entry["answer"])
        correct = int(cleaned == entry["answer"].lower())
        
        # ========== EXTRACT TOKEN CONFIDENCE FEATURES ==========
        
        if hasattr(gen_output, 'scores') and gen_output.scores:
            # scores is a tuple of tensors, one per generated token
            # Each tensor has shape [batch_size, vocab_size]
            
            token_confidences = []
            for score_tensor in gen_output.scores:
                # Get probability distribution
                probs = torch.softmax(score_tensor[0], dim=-1)
                # Get max probability (confidence in chosen token)
                max_prob = probs.max().item()
                token_confidences.append(max_prob)
            
            # Primary features
            avg_confidence = float(np.mean(token_confidences))
            min_confidence = float(np.min(token_confidences))
            max_confidence = float(np.max(token_confidences))
            std_confidence = float(np.std(token_confidences))
            
            # First token is often most important
            first_token_conf = float(token_confidences[0]) if token_confidences else None
            
            # Last token confidence
            last_token_conf = float(token_confidences[-1]) if token_confidences else None
            
            # Confidence drop (early tokens vs late tokens)
            if len(token_confidences) > 1:
                conf_drop = float(token_confidences[0] - token_confidences[-1])
                # Trend: is confidence increasing or decreasing?
                x = np.arange(len(token_confidences))
                conf_trend = float(np.polyfit(x, token_confidences, 1)[0])
            else:
                conf_drop = None
                conf_trend = None
            
            # Geometric mean (better for probabilities)
            # Add small epsilon to avoid log(0)
            token_conf_array = np.array(token_confidences) + 1e-10
            geo_mean_conf = float(np.exp(np.mean(np.log(token_conf_array))))
            
            # Minimum confidence in first 3 tokens (critical region)
            first_3_min = float(np.min(token_confidences[:3])) if len(token_confidences) >= 3 else min_confidence
            
        else:
            # No scores available (shouldn't happen)
            avg_confidence = min_confidence = max_confidence = std_confidence = None
            first_token_conf = last_token_conf = conf_drop = conf_trend = None
            geo_mean_conf = first_3_min = None
        
        # ========== SECONDARY FEATURES ==========
        
        # Answer length
        answer_length = len(gen_output.scores) if hasattr(gen_output, 'scores') else 0
        
        # Store all features
        features.append({
            "question_id": entry["question_id"],
            "category": entry.get("category", "unknown"),
            "answer": entry["answer"],
            "predicted": pred_text,
            "cleaned": cleaned,
            "correct": correct,
            
            # Token confidence features
            "avg_confidence": avg_confidence,
            "min_confidence": min_confidence,
            "max_confidence": max_confidence,
            "std_confidence": std_confidence,
            "first_token_conf": first_token_conf,
            "last_token_conf": last_token_conf,
            "conf_drop": conf_drop,
            "conf_trend": conf_trend,
            "geo_mean_conf": geo_mean_conf,
            "first_3_min": first_3_min,
            
            # Sequence features
            "answer_length": answer_length,
        })
        
        # Clean up
        del inputs, gen_output, image
        torch.cuda.empty_cache()
        
    except Exception as e:
        skipped += 1
        if skipped <= 5:  # Only print first few errors
            print(f"\n⚠️ Skipped sample {i}: {e}")

# Save features
os.makedirs("idea_4/results", exist_ok=True)
output_path = "idea_4/results/token_confidence_features.json"
with open(output_path, "w") as f:
    json.dump(features, f, indent=2)

print(f"\n✅ Saved {len(features)} feature records to {output_path}")
print(f"⚠️ Skipped {skipped} samples due to errors")

# Quick statistics
correct_count = sum(f["correct"] for f in features)
print(f"\n📊 Dataset Statistics:")
print(f"   Total samples: {len(features)}")
print(f"   Correct: {correct_count} ({correct_count/len(features)*100:.1f}%)")
print(f"   Incorrect: {len(features)-correct_count} ({(len(features)-correct_count)/len(features)*100:.1f}%)")

# Check if features are informative
if features:
    correct_confs = [f["avg_confidence"] for f in features if f["correct"] == 1 and f["avg_confidence"] is not None]
    incorrect_confs = [f["avg_confidence"] for f in features if f["correct"] == 0 and f["avg_confidence"] is not None]
    
    if correct_confs and incorrect_confs:
        print(f"\n🔍 Feature Check (avg_confidence):")
        print(f"   Correct answers:   mean={np.mean(correct_confs):.4f}")
        print(f"   Incorrect answers: mean={np.mean(incorrect_confs):.4f}")
        print(f"   Separation: {abs(np.mean(correct_confs) - np.mean(incorrect_confs)):.4f}")
        if abs(np.mean(correct_confs) - np.mean(incorrect_confs)) < 0.01:
            print(f"   ⚠️ WARNING: Features may not be discriminative enough!")
```

```python
# ============================================
# CELL 8 — TRAIN CONFIDENCE CLASSIFIER
# ============================================
import os, json, joblib, numpy as np, pandas as pd
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split, cross_val_score, StratifiedKFold
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, roc_auc_score

RESULTS_DIR = "idea_4/results"
os.makedirs(RESULTS_DIR, exist_ok=True)

# ---------- Load features ----------
print("📂 Loading features...")
with open("idea_4/results/token_confidence_features.json", "r") as f:
    records = json.load(f)

df = pd.DataFrame(records)
print(f"✅ Loaded {len(df)} samples")

# ---------- Data validation ----------
print(f"\n📊 Class Distribution:")
print(f"   Correct (1):   {df['correct'].sum()} samples ({df['correct'].mean()*100:.1f}%)")
print(f"   Incorrect (0): {(1-df['correct']).sum()} samples ({(1-df['correct']).mean()*100:.1f}%)")

# Check for class imbalance
class_ratio = min(df['correct'].sum(), len(df) - df['correct'].sum()) / len(df)
if class_ratio < 0.2:
    print(f"   ⚠️ WARNING: Severe class imbalance (ratio: {class_ratio:.2%})")

# ---------- Feature selection ----------
feature_cols = [
    "avg_confidence", "min_confidence", "max_confidence", "std_confidence",
    "first_token_conf", "last_token_conf", "conf_drop", "conf_trend",
    "geo_mean_conf", "first_3_min", "answer_length"
]

# Remove samples missing critical features
critical_features = ["avg_confidence", "min_confidence", "first_token_conf"]
df_clean = df.dropna(subset=critical_features)
print(f"\n🧹 After removing missing critical features: {len(df_clean)} samples")

if len(df_clean) < 100:
    print(f"❌ ERROR: Only {len(df_clean)} samples remaining!")
    print(f"   Need at least 100 for reliable training.")
    print(f"   Try increasing NUM_SAMPLES in Cell 7 to 3000+")
    raise ValueError("Insufficient data for training")

# Prepare feature matrix
X = df_clean[feature_cols].fillna(0)
y = df_clean["correct"].astype(int)

# Remove infinite values
X = X.replace([np.inf, -np.inf], np.nan).fillna(0)

# ---------- Feature analysis ----------
print("\n" + "="*70)
print("FEATURE ANALYSIS")
print("="*70)

for col in ["avg_confidence", "min_confidence", "first_token_conf", "answer_length"]:
    correct_vals = X.loc[y == 1, col]
    incorrect_vals = X.loc[y == 0, col]
    
    print(f"\n{col}:")
    print(f"  Correct (1):   mean={correct_vals.mean():.4f}, std={correct_vals.std():.4f}")
    print(f"  Incorrect (0): mean={incorrect_vals.mean():.4f}, std={incorrect_vals.std():.4f}")
    print(f"  Difference:    {abs(correct_vals.mean() - incorrect_vals.mean()):.4f}")
    
    # Check if feature is discriminative
    if abs(correct_vals.mean() - incorrect_vals.mean()) < 0.01:
        print(f"  ⚠️ WARNING: Low discrimination power!")

# ---------- Preprocessing ----------
print("\n" + "="*70)
print("PREPROCESSING")
print("="*70)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Stratified split to maintain class balance
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, 
    test_size=0.25, 
    random_state=42, 
    stratify=y
)

print(f"Train set: {len(X_train)} samples")
print(f"Test set:  {len(X_test)} samples")

# ---------- Train multiple models ----------
print("\n" + "="*70)
print("MODEL TRAINING")
print("="*70)

models = {}

# 1. Logistic Regression (fast baseline)
print("\n1️⃣ Logistic Regression")
print("-" * 50)
lr = LogisticRegression(
    max_iter=1000,
    class_weight='balanced',
    random_state=42
)
lr.fit(X_train, y_train)
lr_pred = lr.predict(X_test)
lr_acc = accuracy_score(y_test, lr_pred)
print(f"   Test Accuracy: {lr_acc:.3f}")
models['Logistic'] = (lr, lr_acc)

# Show feature importance
coefs = pd.DataFrame({
    'feature': feature_cols,
    'coefficient': np.abs(lr.coef_[0])
}).sort_values('coefficient', ascending=False)
print("\n   Top 5 Features:")
for idx, row in coefs.head().iterrows():
    print(f"      {row['feature']:<20s}: {row['coefficient']:.4f}")

# 2. Random Forest
print("\n2️⃣ Random Forest")
print("-" * 50)
rf = RandomForestClassifier(
    n_estimators=200,
    max_depth=10,
    min_samples_split=10,
    min_samples_leaf=5,
    max_features='sqrt',
    class_weight='balanced',
    random_state=42,
    n_jobs=-1
)
rf.fit(X_train, y_train)
rf_pred = rf.predict(X_test)
rf_acc = accuracy_score(y_test, rf_pred)
print(f"   Test Accuracy: {rf_acc:.3f}")
models['RandomForest'] = (rf, rf_acc)

# Show feature importance
importances = pd.DataFrame({
    'feature': feature_cols,
    'importance': rf.feature_importances_
}).sort_values('importance', ascending=False)
print("\n   Top 5 Features:")
for idx, row in importances.head().iterrows():
    print(f"      {row['feature']:<20s}: {row['importance']:.4f}")

# 3. Gradient Boosting
print("\n3️⃣ Gradient Boosting")
print("-" * 50)
gb = GradientBoostingClassifier(
    n_estimators=150,
    learning_rate=0.1,
    max_depth=5,
    min_samples_split=10,
    min_samples_leaf=5,
    subsample=0.8,
    random_state=42
)
gb.fit(X_train, y_train)
gb_pred = gb.predict(X_test)
gb_acc = accuracy_score(y_test, gb_pred)
print(f"   Test Accuracy: {gb_acc:.3f}")
models['GradientBoosting'] = (gb, gb_acc)

# ---------- Select best model ----------
best_name = max(models, key=lambda k: models[k][1])
best_clf, best_acc = models[best_name]

print("\n" + "="*70)
print(f"🏆 BEST MODEL: {best_name}")
print("="*70)

# Get predictions from best model
if best_name == 'Logistic':
    best_pred = lr_pred
elif best_name == 'RandomForest':
    best_pred = rf_pred
else:
    best_pred = gb_pred

print(f"\nTest Accuracy: {best_acc:.3f}")
print("\n" + classification_report(y_test, best_pred, digits=3))

# Confusion matrix
cm = confusion_matrix(y_test, best_pred)
print("Confusion Matrix:")
print(f"                  Predicted")
print(f"                 0        1")
print(f"Actual  0     {cm[0,0]:4d}    {cm[0,1]:4d}")
print(f"        1     {cm[1,0]:4d}    {cm[1,1]:4d}")

# Calculate additional metrics
tn, fp, fn, tp = cm.ravel()
precision = tp / (tp + fp) if (tp + fp) > 0 else 0
recall = tp / (tp + fn) if (tp + fn) > 0 else 0
print(f"\nPrecision: {precision:.3f}")
print(f"Recall:    {recall:.3f}")

# Cross-validation
print("\n" + "-"*70)
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_scores = cross_val_score(best_clf, X_scaled, y, cv=cv, scoring='accuracy')
print(f"🔁 5-Fold Cross-Validation: {cv_scores.mean():.3f} ± {cv_scores.std():.3f}")
print(f"   Individual folds: {[f'{s:.3f}' for s in cv_scores]}")

# ---------- Save model ----------
print("\n" + "="*70)
print("SAVING MODEL")
print("="*70)

joblib.dump(best_clf, os.path.join(RESULTS_DIR, "confidence_classifier.pkl"))
joblib.dump(scaler, os.path.join(RESULTS_DIR, "confidence_scaler.pkl"))

metadata = {
    "model_type": best_name,
    "test_accuracy": float(best_acc),
    "cv_accuracy": float(cv_scores.mean()),
    "cv_std": float(cv_scores.std()),
    "features": feature_cols,
    "num_training_samples": len(X_train),
    "num_test_samples": len(X_test),
    "class_balance": {
        "correct": int(y.sum()),
        "incorrect": int(len(y) - y.sum())
    }
}

with open(os.path.join(RESULTS_DIR, "classifier_metadata.json"), "w") as f:
    json.dump(metadata, f, indent=2)

print(f"✅ Saved {best_name} classifier")
print(f"✅ Saved StandardScaler")
print(f"✅ Saved metadata")

# ---------- Final diagnostic ----------
print("\n" + "="*70)
print("DIAGNOSTIC")
print("="*70)

if best_acc < 0.60:
    print("⚠️ Accuracy is below 60%. Possible issues:")
    print("   1. Features may not be discriminative enough")
    print("   2. Model may get many questions wrong (check VQA accuracy)")
    print("   3. Try increasing NUM_SAMPLES to 3000+ in Cell 7")
    print("   4. Check if 'correct' labels are accurate")
elif best_acc < 0.70:
    print("✅ Decent performance. To improve further:")
    print("   1. Increase NUM_SAMPLES to 5000+ in Cell 7")
    print("   2. Add more diverse question categories")
else:
    print("🎉 Great performance! Classifier is working well.")

print(f"\nExpected accuracy if classifier is good: 65-80%")
print(f"Your accuracy: {best_acc:.1%}")
```

```python
# ============================================
# CELL 9 — COMPREHENSIVE VISUALIZATIONS
# ============================================
import json, joblib, numpy as np, pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, roc_curve, auc, precision_recall_curve
from sklearn.model_selection import learning_curve
import os

# Set style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (15, 10)
plt.rcParams['font.size'] = 10

RESULTS_DIR = "idea_4/results"
os.makedirs(f"{RESULTS_DIR}/plots", exist_ok=True)

# ---------- Load data ----------
print("📂 Loading data...")
with open(f"{RESULTS_DIR}/token_confidence_features.json", "r") as f:
    records = json.load(f)

df = pd.DataFrame(records)
clf = joblib.load(f"{RESULTS_DIR}/confidence_classifier.pkl")
scaler = joblib.load(f"{RESULTS_DIR}/confidence_scaler.pkl")

# Prepare features
feature_cols = [
    "avg_confidence", "min_confidence", "max_confidence", "std_confidence",
    "first_token_conf", "last_token_conf", "conf_drop", "conf_trend",
    "geo_mean_conf", "first_3_min", "answer_length"
]

df_clean = df.dropna(subset=["avg_confidence", "min_confidence", "first_token_conf"])
X = df_clean[feature_cols].fillna(0).replace([np.inf, -np.inf], np.nan).fillna(0)
y = df_clean["correct"].astype(int)
X_scaled = scaler.transform(X)

print(f"✅ Loaded {len(df_clean)} samples")

# ---------- CREATE COMPREHENSIVE PLOT ----------
fig = plt.figure(figsize=(20, 16))
gs = fig.add_gridspec(4, 3, hspace=0.3, wspace=0.3)

# ========== 1. FEATURE DISTRIBUTIONS (Top Row) ==========
print("📊 Creating feature distribution plots...")

key_features = ["avg_confidence", "min_confidence", "first_token_conf"]
for idx, feature in enumerate(key_features):
    ax = fig.add_subplot(gs[0, idx])
    
    correct_data = df_clean[df_clean["correct"] == 1][feature].dropna()
    incorrect_data = df_clean[df_clean["correct"] == 0][feature].dropna()
    
    # Overlapping histograms
    ax.hist(correct_data, bins=30, alpha=0.6, label='Correct', color='green', edgecolor='black')
    ax.hist(incorrect_data, bins=30, alpha=0.6, label='Incorrect', color='red', edgecolor='black')
    
    ax.set_xlabel(feature.replace('_', ' ').title())
    ax.set_ylabel('Frequency')
    ax.set_title(f'{feature.replace("_", " ").title()} Distribution')
    ax.legend()
    ax.grid(True, alpha=0.3)

# ========== 2. CONFUSION MATRIX (Row 2, Col 1) ==========
print("📊 Creating confusion matrix...")
ax = fig.add_subplot(gs[1, 0])

y_pred = clf.predict(X_scaled)
cm = confusion_matrix(y, y_pred)

sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax, 
            xticklabels=['Incorrect', 'Correct'],
            yticklabels=['Incorrect', 'Correct'],
            cbar_kws={'label': 'Count'})
ax.set_ylabel('Actual')
ax.set_xlabel('Predicted')
ax.set_title('Confusion Matrix')

# Add accuracy text
accuracy = (cm[0,0] + cm[1,1]) / cm.sum()
ax.text(1, -0.15, f'Accuracy: {accuracy:.1%}', 
        transform=ax.transAxes, ha='center', fontsize=12, weight='bold')

# ========== 3. ROC CURVE (Row 2, Col 2) ==========
print("📊 Creating ROC curve...")
ax = fig.add_subplot(gs[1, 1])

# Get prediction probabilities
y_proba = clf.predict_proba(X_scaled)[:, 1]
fpr, tpr, thresholds = roc_curve(y, y_proba)
roc_auc = auc(fpr, tpr)

ax.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (AUC = {roc_auc:.3f})')
ax.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--', label='Random Classifier')
ax.set_xlim([0.0, 1.0])
ax.set_ylim([0.0, 1.05])
ax.set_xlabel('False Positive Rate')
ax.set_ylabel('True Positive Rate')
ax.set_title('ROC Curve')
ax.legend(loc="lower right")
ax.grid(True, alpha=0.3)

# ========== 4. PRECISION-RECALL CURVE (Row 2, Col 3) ==========
print("📊 Creating precision-recall curve...")
ax = fig.add_subplot(gs[1, 2])

precision, recall, _ = precision_recall_curve(y, y_proba)
pr_auc = auc(recall, precision)

ax.plot(recall, precision, color='purple', lw=2, label=f'PR curve (AUC = {pr_auc:.3f})')
ax.set_xlim([0.0, 1.0])
ax.set_ylim([0.0, 1.05])
ax.set_xlabel('Recall')
ax.set_ylabel('Precision')
ax.set_title('Precision-Recall Curve')
ax.legend(loc="lower left")
ax.grid(True, alpha=0.3)

# ========== 5. FEATURE IMPORTANCE (Row 3, Col 1) ==========
print("📊 Creating feature importance plot...")
ax = fig.add_subplot(gs[2, 0])

if hasattr(clf, 'feature_importances_'):
    importances = clf.feature_importances_
elif hasattr(clf, 'coef_'):
    importances = np.abs(clf.coef_[0])
else:
    importances = np.zeros(len(feature_cols))

feat_imp = pd.DataFrame({
    'feature': feature_cols,
    'importance': importances
}).sort_values('importance', ascending=True)

ax.barh(feat_imp['feature'], feat_imp['importance'], color='steelblue', edgecolor='black')
ax.set_xlabel('Importance')
ax.set_title('Feature Importance')
ax.grid(True, alpha=0.3, axis='x')

# ========== 6. CONFIDENCE VS CORRECTNESS (Row 3, Col 2) ==========
print("📊 Creating confidence scatter plot...")
ax = fig.add_subplot(gs[2, 1])

correct_df = df_clean[df_clean["correct"] == 1]
incorrect_df = df_clean[df_clean["correct"] == 0]

ax.scatter(correct_df["avg_confidence"], correct_df["min_confidence"], 
           alpha=0.6, s=50, label='Correct', color='green', edgecolors='black', linewidth=0.5)
ax.scatter(incorrect_df["avg_confidence"], incorrect_df["min_confidence"], 
           alpha=0.6, s=50, label='Incorrect', color='red', edgecolors='black', linewidth=0.5)

ax.set_xlabel('Average Confidence')
ax.set_ylabel('Minimum Confidence')
ax.set_title('Confidence Scatter: Avg vs Min')
ax.legend()
ax.grid(True, alpha=0.3)

# ========== 7. ANSWER LENGTH DISTRIBUTION (Row 3, Col 3) ==========
print("📊 Creating answer length distribution...")
ax = fig.add_subplot(gs[2, 2])

correct_lengths = df_clean[df_clean["correct"] == 1]["answer_length"].dropna()
incorrect_lengths = df_clean[df_clean["correct"] == 0]["answer_length"].dropna()

ax.hist(correct_lengths, bins=range(0, 25), alpha=0.6, label='Correct', 
        color='green', edgecolor='black')
ax.hist(incorrect_lengths, bins=range(0, 25), alpha=0.6, label='Incorrect', 
        color='red', edgecolor='black')

ax.set_xlabel('Answer Length (tokens)')
ax.set_ylabel('Frequency')
ax.set_title('Answer Length Distribution')
ax.legend()
ax.grid(True, alpha=0.3)

# ========== 8. CATEGORY PERFORMANCE (Row 4, spans all columns) ==========
print("📊 Creating category performance...")
ax = fig.add_subplot(gs[3, :])

category_stats = df_clean.groupby('category').agg({
    'correct': ['sum', 'count', 'mean']
}).reset_index()
category_stats.columns = ['category', 'correct_count', 'total', 'accuracy']
category_stats = category_stats.sort_values('accuracy', ascending=True)

# Filter categories with at least 10 samples
category_stats = category_stats[category_stats['total'] >= 10]

colors = ['red' if acc < 0.5 else 'orange' if acc < 0.65 else 'green' 
          for acc in category_stats['accuracy']]

bars = ax.barh(category_stats['category'], category_stats['accuracy'], color=colors, 
               edgecolor='black', linewidth=1)

# Add count labels
for idx, (cat, acc, total) in enumerate(zip(category_stats['category'], 
                                              category_stats['accuracy'], 
                                              category_stats['total'])):
    ax.text(acc + 0.02, idx, f'n={int(total)}', va='center', fontsize=9)

ax.set_xlabel('Accuracy')
ax.set_title('Model Accuracy by Question Category (min 10 samples)')
ax.set_xlim([0, 1.0])
ax.axvline(x=0.5, color='black', linestyle='--', linewidth=1, alpha=0.5, label='Random')
ax.legend()
ax.grid(True, alpha=0.3, axis='x')

# ========== SAVE FIGURE ==========
plt.suptitle('VQA Confidence Classifier - Comprehensive Analysis', 
             fontsize=16, weight='bold', y=0.995)

output_path = f"{RESULTS_DIR}/plots/comprehensive_analysis.png"
plt.savefig(output_path, dpi=300, bbox_inches='tight')
print(f"\n✅ Saved comprehensive plot to {output_path}")

plt.show()

# ========== PRINT SUMMARY STATISTICS ==========
print("\n" + "="*70)
print("SUMMARY STATISTICS")
print("="*70)

print(f"\nOverall Performance:")
print(f"  Accuracy:  {accuracy:.3f}")
print(f"  ROC AUC:   {roc_auc:.3f}")
print(f"  PR AUC:    {pr_auc:.3f}")

print(f"\nConfusion Matrix:")
tn, fp, fn, tp = cm.ravel()
print(f"  True Negatives:  {tn:4d}   False Positives: {fp:4d}")
print(f"  False Negatives: {fn:4d}   True Positives:  {tp:4d}")

precision = tp / (tp + fp) if (tp + fp) > 0 else 0
recall = tp / (tp + fn) if (tp + fn) > 0 else 0
f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

print(f"\nMetrics:")
print(f"  Precision: {precision:.3f}")
print(f"  Recall:    {recall:.3f}")
print(f"  F1-Score:  {f1:.3f}")

print(f"\nFeature Statistics:")
for feat in ["avg_confidence", "min_confidence", "first_token_conf"]:
    correct_mean = df_clean[df_clean["correct"] == 1][feat].mean()
    incorrect_mean = df_clean[df_clean["correct"] == 0][feat].mean()
    print(f"  {feat}:")
    print(f"    Correct:   {correct_mean:.4f}")
    print(f"    Incorrect: {incorrect_mean:.4f}")
    print(f"    Gap:       {abs(correct_mean - incorrect_mean):.4f}")

print("\n" + "="*70)
```

```python
# ============================================
# CELL 10 — INDIVIDUAL DIAGNOSTIC PLOTS
# ============================================
import json, joblib, numpy as np, pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import os

RESULTS_DIR = "idea_4/results"

# Load data
with open(f"{RESULTS_DIR}/token_confidence_features.json", "r") as f:
    records = json.load(f)

df = pd.DataFrame(records)
df_clean = df.dropna(subset=["avg_confidence", "min_confidence", "first_token_conf"])

print(f"📊 Creating individual diagnostic plots...")

# ========== PLOT 1: Confidence Distribution Box Plots ==========
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
fig.suptitle('Confidence Features: Correct vs Incorrect Answers', fontsize=16, weight='bold')

confidence_features = [
    "avg_confidence", "min_confidence", "max_confidence",
    "first_token_conf", "geo_mean_conf", "first_3_min"
]

for idx, feat in enumerate(confidence_features):
    ax = axes[idx // 3, idx % 3]
    
    data_to_plot = [
        df_clean[df_clean["correct"] == 0][feat].dropna(),
        df_clean[df_clean["correct"] == 1][feat].dropna()
    ]
    
    bp = ax.boxplot(data_to_plot, labels=['Incorrect', 'Correct'], 
                     patch_artist=True, notch=True)
    
    # Color the boxes
    bp['boxes'][0].set_facecolor('lightcoral')
    bp['boxes'][1].set_facecolor('lightgreen')
    
    ax.set_ylabel('Value')
    ax.set_title(feat.replace('_', ' ').title())
    ax.grid(True, alpha=0.3, axis='y')
    
    # Add mean markers
    means = [np.mean(d) for d in data_to_plot]
    ax.plot([1, 2], means, 'D', color='blue', markersize=8, label='Mean')
    
    # Statistical test
    if len(data_to_plot[0]) > 0 and len(data_to_plot[1]) > 0:
        t_stat, p_val = stats.ttest_ind(data_to_plot[0], data_to_plot[1])
        sig = "***" if p_val < 0.001 else "**" if p_val < 0.01 else "*" if p_val < 0.05 else "ns"
        ax.text(0.5, 0.95, f'p={p_val:.4f} {sig}', 
                transform=ax.transAxes, ha='center', va='top',
                bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

plt.tight_layout()
plt.savefig(f"{RESULTS_DIR}/plots/confidence_boxplots.png", dpi=300, bbox_inches='tight')
print(f"✅ Saved confidence boxplots")
plt.show()

# ========== PLOT 2: Confidence Trajectories ==========
print("📊 Creating confidence trajectory plot...")

# Sample some examples to visualize
n_examples = 20
sample_df = df_clean.sample(min(n_examples, len(df_clean)), random_state=42)

fig, ax = plt.subplots(figsize=(14, 8))

for _, row in sample_df.iterrows():
    # We don't have per-token data, so we'll simulate trajectory with available features
    if pd.notna(row["first_token_conf"]) and pd.notna(row["last_token_conf"]):
        length = int(row["answer_length"]) if pd.notna(row["answer_length"]) else 5
        
        # Interpolate between first and last
        if length > 1:
            trajectory = np.linspace(row["first_token_conf"], row["last_token_conf"], length)
            color = 'green' if row["correct"] == 1 else 'red'
            alpha = 0.6
            ax.plot(range(length), trajectory, color=color, alpha=alpha, linewidth=2)

ax.set_xlabel('Token Position', fontsize=12)
ax.set_ylabel('Confidence', fontsize=12)
ax.set_title('Token Confidence Trajectories (Sample)', fontsize=14, weight='bold')
ax.grid(True, alpha=0.3)

# Add legend
from matplotlib.lines import Line2D
legend_elements = [
    Line2D([0], [0], color='green', lw=2, label='Correct'),
    Line2D([0], [0], color='red', lw=2, label='Incorrect')
]
ax.legend(handles=legend_elements, loc='best')

plt.tight_layout()
plt.savefig(f"{RESULTS_DIR}/plots/confidence_trajectories.png", dpi=300, bbox_inches='tight')
print(f"✅ Saved confidence trajectories")
plt.show()

# ========== PLOT 3: 2D Feature Space ==========
print("📊 Creating 2D feature space visualization...")

fig, axes = plt.subplots(2, 2, figsize=(16, 14))
fig.suptitle('2D Feature Space Visualization', fontsize=16, weight='bold')

feature_pairs = [
    ("avg_confidence", "min_confidence"),
    ("first_token_conf", "answer_length"),
    ("avg_confidence", "std_confidence"),
    ("geo_mean_conf", "conf_drop")
]

for idx, (feat_x, feat_y) in enumerate(feature_pairs):
    ax = axes[idx // 2, idx % 2]
    
    correct_df = df_clean[df_clean["correct"] == 1]
    incorrect_df = df_clean[df_clean["correct"] == 0]
    
    ax.scatter(incorrect_df[feat_x], incorrect_df[feat_y], 
               alpha=0.5, s=60, label='Incorrect', color='red', 
               edgecolors='darkred', linewidth=0.5)
    ax.scatter(correct_df[feat_x], correct_df[feat_y], 
               alpha=0.5, s=60, label='Correct', color='green', 
               edgecolors='darkgreen', linewidth=0.5)
    
    ax.set_xlabel(feat_x.replace('_', ' ').title(), fontsize=11)
    ax.set_ylabel(feat_y.replace('_', ' ').title(), fontsize=11)
    ax.set_title(f'{feat_x.replace("_", " ").title()} vs {feat_y.replace("_", " ").title()}')
    ax.legend()
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig(f"{RESULTS_DIR}/plots/2d_feature_space.png", dpi=300, bbox_inches='tight')
print(f"✅ Saved 2D feature space")
plt.show()

# ========== PLOT 4: Prediction Confidence Distribution ==========
print("📊 Creating prediction confidence distribution...")

clf = joblib.load(f"{RESULTS_DIR}/confidence_classifier.pkl")
scaler = joblib.load(f"{RESULTS_DIR}/confidence_scaler.pkl")

feature_cols = [
    "avg_confidence", "min_confidence", "max_confidence", "std_confidence",
    "first_token_conf", "last_token_conf", "conf_drop", "conf_trend",
    "geo_mean_conf", "first_3_min", "answer_length"
]

X = df_clean[feature_cols].fillna(0).replace([np.inf, -np.inf], np.nan).fillna(0)
y = df_clean["correct"].astype(int)
X_scaled = scaler.transform(X)

# Get prediction probabilities
y_proba = clf.predict_proba(X_scaled)[:, 1]

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Left: Histogram of prediction probabilities
ax = axes[0]
correct_proba = y_proba[y == 1]
incorrect_proba = y_proba[y == 0]

ax.hist(incorrect_proba, bins=30, alpha=0.6, label='Actually Incorrect', 
        color='red', edgecolor='black')
ax.hist(correct_proba, bins=30, alpha=0.6, label='Actually Correct', 
        color='green', edgecolor='black')

ax.axvline(x=0.5, color='black', linestyle='--', linewidth=2, label='Decision Threshold')
ax.set_xlabel('Predicted Probability (Correct)', fontsize=12)
ax.set_ylabel('Frequency', fontsize=12)
ax.set_title('Classifier Prediction Confidence Distribution', fontsize=14, weight='bold')
ax.legend()
ax.grid(True, alpha=0.3)

# Right: Calibration plot
ax = axes[1]

# Bin predictions
bins = np.linspace(0, 1, 11)
bin_centers = (bins[:-1] + bins[1:]) / 2
bin_accuracies = []
bin_counts = []

for i in range(len(bins) - 1):
    mask = (y_proba >= bins[i]) & (y_proba < bins[i+1])
    if mask.sum() > 0:
        bin_accuracies.append(y[mask].mean())
        bin_counts.append(mask.sum())
    else:
        bin_accuracies.append(np.nan)
        bin_counts.append(0)

ax.plot([0, 1], [0, 1], 'k--', linewidth=2, label='Perfect Calibration')
ax.plot(bin_centers, bin_accuracies, 'o-', linewidth=2, markersize=8, 
        color='blue', label='Actual Calibration')

# Add count annotations
for x, y_val, count in zip(bin_centers, bin_accuracies, bin_counts):
    if not np.isnan(y_val) and count > 0:
        ax.annotate(f'n={count}', (x, y_val), textcoords="offset points", 
                   xytext=(0, 10), ha='center', fontsize=8)

ax.set_xlabel('Predicted Probability', fontsize=12)
ax.set_ylabel('Actual Accuracy', fontsize=12)
ax.set_title('Classifier Calibration Curve', fontsize=14, weight='bold')
ax.legend()
ax.grid(True, alpha=0.3)
ax.set_xlim([0, 1])
ax.set_ylim([0, 1])

plt.tight_layout()
plt.savefig(f"{RESULTS_DIR}/plots/prediction_confidence.png", dpi=300, bbox_inches='tight')
print(f"✅ Saved prediction confidence plots")
plt.show()

# ========== PLOT 5: Error Analysis ==========
print("📊 Creating error analysis...")

y_pred = clf.predict(X_scaled)
errors = (y_pred != y)

fig, axes = plt.subplots(2, 2, figsize=(16, 14))
fig.suptitle('Error Analysis', fontsize=16, weight='bold')

# Top-left: False Positives vs True Positives
ax = axes[0, 0]
tp = (y == 1) & (y_pred == 1)
fp = (y == 0) & (y_pred == 1)

ax.scatter(df_clean[tp]["avg_confidence"], df_clean[tp]["min_confidence"],
           alpha=0.6, s=60, label='True Positive', color='darkgreen', edgecolors='black')
ax.scatter(df_clean[fp]["avg_confidence"], df_clean[fp]["min_confidence"],
           alpha=0.6, s=60, label='False Positive', color='orange', edgecolors='black')
ax.set_xlabel('Average Confidence')
ax.set_ylabel('Minimum Confidence')
ax.set_title('False Positives vs True Positives')
ax.legend()
ax.grid(True, alpha=0.3)

# Top-right: False Negatives vs True Negatives
ax = axes[0, 1]
tn = (y == 0) & (y_pred == 0)
fn = (y == 1) & (y_pred == 0)

ax.scatter(df_clean[tn]["avg_confidence"], df_clean[tn]["min_confidence"],
           alpha=0.6, s=60, label='True Negative', color='darkred', edgecolors='black')
ax.scatter(df_clean[fn]["avg_confidence"], df_clean[fn]["min_confidence"],
           alpha=0.6, s=60, label='False Negative', color='purple', edgecolors='black')
ax.set_xlabel('Average Confidence')
ax.set_ylabel('Minimum Confidence')
ax.set_title('False Negatives vs True Negatives')
ax.legend()
ax.grid(True, alpha=0.3)

# Bottom-left: Error rate by confidence bucket
ax = axes[1, 0]
conf_bins = np.linspace(0, 1, 11)
bin_labels = [f'{conf_bins[i]:.1f}-{conf_bins[i+1]:.1f}' for i in range(len(conf_bins)-1)]
bin_errors = []

for i in range(len(conf_bins) - 1):
    mask = (df_clean["avg_confidence"] >= conf_bins[i]) & (df_clean["avg_confidence"] < conf_bins[i+1])
    if mask.sum() > 0:
        error_rate = errors[mask].mean()
        bin_errors.append(error_rate)
    else:
        bin_errors.append(0)

ax.bar(range(len(bin_errors)), bin_errors, color='coral', edgecolor='black')
ax.set_xticks(range(len(bin_errors)))
ax.set_xticklabels(bin_labels, rotation=45, ha='right')
ax.set_xlabel('Average Confidence Bucket')
ax.set_ylabel('Error Rate')
ax.set_title('Classification Error Rate by Confidence Level')
ax.grid(True, alpha=0.3, axis='y')

# Bottom-right: Category error rates
ax = axes[1, 1]
category_errors = df_clean.groupby('category').apply(
    lambda x: (clf.predict(scaler.transform(x[feature_cols].fillna(0).replace([np.inf, -np.inf], np.nan).fillna(0))) 
               != x['correct']).mean()
).sort_values(ascending=False)

# Only show categories with at least 10 samples
category_counts = df_clean['category'].value_counts()
category_errors = category_errors[category_counts[category_errors.index] >= 10]

colors = ['darkred' if err > 0.5 else 'orange' if err > 0.35 else 'lightgreen' for err in category_errors]
ax.barh(category_errors.index, category_errors.values, color=colors, edgecolor='black')
ax.set_xlabel('Error Rate')
ax.set_title('Classification Error Rate by Category (min 10 samples)')
ax.axvline(x=0.5, color='black', linestyle='--', linewidth=1, alpha=0.5)
ax.grid(True, alpha=0.3, axis='x')

plt.tight_layout()
plt.savefig(f"{RESULTS_DIR}/plots/error_analysis.png", dpi=300, bbox_inches='tight')
print(f"✅ Saved error analysis")
plt.show()

print("\n" + "="*70)
print("✅ ALL PLOTS GENERATED SUCCESSFULLY")
print("="*70)
print(f"\nPlots saved in: {RESULTS_DIR}/plots/")
print(f"  1. comprehensive_analysis.png")
print(f"  2. confidence_boxplots.png")
print(f"  3. confidence_trajectories.png")
print(f"  4. 2d_feature_space.png")
print(f"  5. prediction_confidence.png")
print(f"  6. error_analysis.png")
```
