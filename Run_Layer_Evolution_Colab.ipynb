## Step 5: Load and Inspect Results


```python
import json
import pandas as pd

# Load results
with open('layer_evolution_results.json', 'r') as f:
    data = json.load(f)

print("Metadata:")
print(json.dumps(data['metadata'], indent=2))

# Convert results to DataFrame for easy analysis
df = pd.DataFrame(data['results'])
print(f"\nProcessed {len(df)} images")
print("\nFirst few results:")
df.head()
```

## Step 6: Basic Statistics


```python
# Statistics on confidence and head delta
print("Token Confidence Statistics:")
print(df['token_confidence'].describe())

print("\nHead Delta Statistics:")
print(df['head_delta'].describe())

print("\nNumber of Generated Tokens Statistics:")
print(df['num_generated_tokens'].describe())
```

## Step 7: Visualize Results


```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Token confidence distribution
axes[0, 0].hist(df['token_confidence'], bins=50, edgecolor='black')
axes[0, 0].set_xlabel('Token Confidence')
axes[0, 0].set_ylabel('Frequency')
axes[0, 0].set_title('Distribution of Token Confidence')

# Head delta distribution
axes[0, 1].hist(df['head_delta'], bins=50, edgecolor='black', color='orange')
axes[0, 1].set_xlabel('Head Delta')
axes[0, 1].set_ylabel('Frequency')
axes[0, 1].set_title('Distribution of Head Delta')

# Number of tokens generated
axes[1, 0].hist(df['num_generated_tokens'], bins=30, edgecolor='black', color='green')
axes[1, 0].set_xlabel('Number of Generated Tokens')
axes[1, 0].set_ylabel('Frequency')
axes[1, 0].set_title('Distribution of Generated Tokens')

# Scatter: confidence vs head_delta
axes[1, 1].scatter(df['token_confidence'], df['head_delta'], alpha=0.5)
axes[1, 1].set_xlabel('Token Confidence')
axes[1, 1].set_ylabel('Head Delta')
axes[1, 1].set_title('Token Confidence vs Head Delta')

plt.tight_layout()
plt.savefig('layer_evolution_analysis.png', dpi=300)
plt.show()

print("\n✅ Visualization saved to: layer_evolution_analysis.png")
```
